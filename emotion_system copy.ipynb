```
train_images = os.path.join(dataset_path, "train", "images")
train_labels = os.path.join(dataset_path, "train", "labels")

val_images = os.path.join(dataset_path, "valid", "images")
val_labels = os.path.join(dataset_path, "valid", "labels")

print("\nTraining Images Folder :", train_images)
print("Training Labels Folder :", train_labels)

print("\nValidation Images Folder :", val_images)
print("Validation Labels Folder :", val_labels)
```

```
# ============================
# 2. COUNT DATASET IMAGES
# ============================

image_extensions = (".jpg", ".jpeg", ".png", ".bmp")

train_count = len([f for f in os.listdir(train_images) if f.endswith(image_extensions)])
val_count = len([f for f in os.listdir(val_images) if f.endswith(image_extensions)])

print("\n------ Dataset Count ------")
print("Training Images :", train_count)
print("Validation Images :", val_count)
print("Total Images :", train_count + val_count)
```

```
# ============================
# 3. IMAGE ANALYSIS
# ============================

print("\n------ Sample Image Analysis ------")

sample_image = None

for img in os.listdir(train_images):
    if img.endswith(image_extensions):
        sample_image = os.path.join(train_images, img)
        break

image = Image.open(sample_image)

print("Image Name :", os.path.basename(sample_image))
print("Image Format :", image.format)
print("Width :", image.width)
print("Height :", image.height)

channels = len(image.getbands())
print("Number of Channels :", channels)
```

```
# ============================
# 4. YOLO LABEL ANALYSIS
# ============================

label_files = [f for f in os.listdir(train_labels) if f.endswith(".txt")]

num_annotations = 0
class_ids = []

for file in label_files:
    with open(os.path.join(train_labels, file), "r") as f:
        lines = f.readlines()
        num_annotations += len(lines)

        for line in lines:
            if line.strip():
                class_id = int(line.split()[0])
                class_ids.append(class_id)

print("\n------ Label Analysis ------")
print("Number of Label Files :", len(label_files))
print("Total Annotations :", num_annotations)
print("Class IDs Found :", sorted(set(class_ids)))
```

```
# ============================
# 5. EMOTION CLASS MAPPING
# ============================

yaml_file = os.path.join(dataset_path, "data.yaml")

if os.path.exists(yaml_file):
    with open(yaml_file, "r") as f:
        data = yaml.safe_load(f)

    print("\n------ Emotion Class Mapping ------")

    names = data["names"]

    if isinstance(names, dict):
        for k, v in names.items():
            print(f"{k} --> {v}")

    elif isinstance(names, list):
        for i, name in enumerate(names):
            print(f"{i} --> {name}")

else:
    print("\ndata.yaml not found.")
```
